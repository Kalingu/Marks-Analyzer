Marks analyzer

```python
# Analyzing students' pass/fail status based on their marks.
# Calculating total, average, minimum, and maximum marks.
```

```python
marks = [65, 80, 45, 90, 70]
total=sum(marks)
print("total:",total)
average=total/len(marks)
print("average:",average)
minimum=min(marks)
print("minimum:",minimum)
highest=max(marks)
print("highest:",highest)
```

```python
for x in marks:
    if x>= 50:
        print(x,"pass")
    else:
        print(x,"fail")
```

```python
def analyze(marks):
    count=0
    for x in marks:
        if x>=50:
            count=count+1
    return count
print(analyze([65, 80, 45, 90, 70]))
```
